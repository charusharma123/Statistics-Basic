# Statistics Basic:

1. What is Statistics, and why it is important?
- Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data. It's important because it helps us make sense of complex data, identify patterns, make informed decisions, test hypotheses, and draw reliable conclusions from observations.

2. What are the two main types of statistics?
- The two main types are:
- Descriptive statistics: Summarize and describe characteristics of data
- Inferential statistics: Make predictions or inferences about a population based on sample data

3. What are descriptive statistics?
- Descriptive statistics are methods used to summarize and describe the main features of a dataset, including measures of central tendency (mean, median, mode), measures of variability (range, variance, standard deviation), and graphical representations (histograms, box plots).

4. What is inferential statistics?
- Inferential statistics involves using sample data to make generalizations or predictions about a larger population. It includes techniques like hypothesis testing, confidence intervals, and regression analysis.

5. What is sampling in statistics?
- Sampling is the process of selecting a subset of individuals or observations from a larger population to estimate characteristics of the whole population.

6. What are the different types of sampling methods?
- Common sampling methods include:
  * Random sampling
  * Stratified sampling
  * Cluster sampling
  * Systematic sampling
  * Convenience sampling
  * Purposive sampling

7. What is the difference between random and non-random sampling?
- Random sampling gives every member of the population an equal chance of being selected (e.g., lottery method), while non-random sampling doesn't use random selection (e.g., convenience sampling where samples are chosen based on availability).

8. Define and give examples of qualitative and quantitative data.
- Qualitative data: Non-numerical data that describes qualities (e.g., colors, emotions, brands)
- Quantitative data: Numerical data that can be measured (e.g., height, weight, temperature)

9. What are the different types of data in statistics?
- Main types are:
  * Nominal (categories without order)
  * Ordinal (ordered categories)
  * Interval (ordered with equal intervals)
  * Ratio (interval with true zero)

10. Explain nominal, ordinal, interval, and ratio levels of measurement.
- Nominal: Categories without order (e.g., gender, colors)
- Ordinal: Ordered categories without defined intervals (e.g., satisfaction levels)
- Interval: Ordered with equal intervals but no true zero (e.g., temperature in Celsius)
- Ratio: Interval with true zero (e.g., height, weight)

11. What is the measure of central tendency?
- Measures of central tendency are values that describe the center of a data distribution, including mean, median, and mode.

## 12. Define mean, median, and mode.
- Mean: The average (sum of values divided by count)
- Median: The middle value when data is ordered
- Mode: The most frequently occurring value

13. What is the significance of the measure of central tendency?
- They provide a single value that represents the center of the data distribution, helping to summarize large datasets and facilitate comparisons between different datasets.

14. What is variance, and how is it calculated?
- Variance measures how far each number in a dataset is from the mean. It's calculated as the average of the squared differences from the mean.

15. What is standard deviation, and why is it important?
- Standard deviation is the square root of variance. It measures dispersion in the same units as the original data and is important because it shows how much variation exists from the average.

16. Define and explain the term range in statistics.
- Range is the difference between the highest and lowest values in a dataset. It's a simple measure of variability.

17. What is the difference between variance and standard deviation?
- Variance is the average squared deviation from the mean, while standard deviation is the square root of variance. Standard deviation is in the same units as the original data.

18. What is skewness in a dataset?
- Skewness measures the asymmetry of the probability distribution of a real-valued random variable about its mean.

19. What does it mean if a dataset is positively or negatively skewed?
- Positively skewed: The tail on the right side is longer (mean > median)
- Negatively skewed: The tail on the left side is longer (mean < median)

20. Define and explain kurtosis.
- Kurtosis measures the "tailedness" of the probability distribution. It indicates whether data are peaked or flat compared to a normal distribution.

21. What is the purpose of covariance?
- Covariance measures how much two random variables change together. It indicates the direction of the linear relationship between variables.

22. What does correlation measure in statistics?
- Correlation measures the strength and direction of the linear relationship between two variables, ranging from -1 to +1.

23. What is the difference between covariance and correlation?
- Covariance indicates direction but not strength of relationship, while correlation standardizes covariance to a range of -1 to +1, showing both direction and strength.

24. What are some real-world applications of statistics?
- Statistics is used in:
  * Business (market research, quality control)
  * Medicine (clinical trials, epidemiology)
  * Economics (forecasting, policy analysis)
  * Sports (player performance analysis)
  * Weather forecasting
  * Political polling
  * Machine learning and AI

9. Write a Python script to calculate convariance between two datasets.

```python
def covariance(x, y):
    n = len(x)
    return sum((xi - np.mean(x)) * (yi - np.mean(y)) for xi, yi in zip(x, y)) / (n - 1)

x = [1, 2, 3, 4, 5]
y = [2, 3, 4, 5, 6]
print(f"Covariance: {covariance(x, y):.2f}")

# Using numpy
print(f"NumPy covariance: {np.cov(x, y)[0, 1]:.2f}")

```

18. Implement Z-score normalization and explain its significance.

```python
def z_score_normalization(data):
    mean = np.mean(data)
    std = np.std(data, ddof=1)
    return [(x - mean)/std for x in data]

normalized = z_score_normalization([1, 2, 3, 4, 5])
print("Normalized data:", normalized)

```

19. Compare two datasets using their standerd deviations.

```python
import numpy as np
import matplotlib.pyplot as plt

# Example datasets (replace with your data)
dataset_A = np.random.normal(loc=50, scale=5, size=1000)  # Mean=50, SD=5
dataset_B = np.random.normal(loc=50, scale=15, size=1000) # Mean=50, SD=15

# Calculate statistics
def compute_stats(data, name):
    mean = np.mean(data)
    std_dev = np.std(data, ddof=1)  # Sample SD (ddof=1)
    cv = (std_dev / mean) * 100  # Coefficient of Variation (%)
    print(f"{name}:")
    print(f"  Mean = {mean:.2f}")
    print(f"  Standard Deviation = {std_dev:.2f}")
    print(f"  Coefficient of Variation (CV) = {cv:.2f}%")
    print("-" * 40)
    return mean, std_dev

mean_A, sd_A = compute_stats(dataset_A, "Dataset A")
mean_B, sd_B = compute_stats(dataset_B, "Dataset B")

# Compare SDs
if sd_A > sd_B:
    print("Dataset A has HIGHER variability than Dataset B.")
elif sd_A < sd_B:
    print("Dataset B has HIGHER variability than Dataset A.")
else:
    print("Both datasets have SIMILAR variability.")

# Plot histograms
plt.figure(figsize=(10, 6))
plt.hist(dataset_A, bins=30, alpha=0.5, label=f"Dataset A (SD={sd_A:.2f})", color='blue')
plt.hist(dataset_B, bins=30, alpha=0.5, label=f"Dataset B (SD={sd_B:.2f})", color='red')
plt.axvline(mean_A, color='blue', linestyle='dashed', linewidth=2, label=f'Mean A: {mean_A:.2f}')
plt.axvline(mean_B, color='red', linestyle='dashed', linewidth=2, label=f'Mean B: {mean_B:.2f}')
plt.title("Comparison of Datasets using Standard Deviation")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

22. Generate a dataset and implement both variance and standard daviation computations.

```python
import random

# Step 1: Generate a dataset of 20 random integers between 10 and 100
data = [random.randint(10, 100) for _ in range(20)]
print("Dataset:", data)

# Step 2: Calculate the mean
mean = sum(data) / len(data)

# Step 3: Calculate the variance (Population)
variance = sum((x - mean) ** 2 for x in data) / len(data)

# Step 4: Calculate the standard deviation (sqrt of variance)
standard_deviation = variance ** 0.5

# Step 5: Display results
print(f"Mean: {mean:.2f}")
print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {standard_deviation:.2f}")
```

24. Implement the Pearson and Spearman correlation coefficients for a dataset.

```python
import numpy as np
from scipy.stats import spearmanr, pearsonr, rankdata

# Step 1: Create two related datasets
x = np.array([10, 20, 30, 40, 50, 60])
y = np.array([12, 24, 33, 47, 55, 68])

# Step 2: Manual Pearson Correlation
def pearson_corr(x, y):
    mean_x, mean_y = np.mean(x), np.mean(y)
    numerator = sum((x - mean_x) * (y - mean_y))
    denominator = np.sqrt(sum((x - mean_x) ** 2) * sum((y - mean_y) ** 2))
    return numerator / denominator

# Step 3: Manual Spearman Correlation
def spearman_corr(x, y):
    rank_x = rankdata(x)
    rank_y = rankdata(y)
    return pearson_corr(rank_x, rank_y)

# Step 4: Calculate both correlations
pearson_manual = pearson_corr(x, y)
spearman_manual = spearman_corr(x, y)

# Step 5: Compare with scipy's built-in functions
pearson_builtin, _ = pearsonr(x, y)
spearman_builtin, _ = spearmanr(x, y)

# Step 6: Display results
print("Manual Pearson Correlation:", round(pearson_manual, 4))
print("Manual Spearman Correlation:", round(spearman_manual, 4))
print("SciPy Pearson Correlation  :", round(pearson_builtin, 4))
print("SciPy Spearman Correlation :", round(spearman_builtin, 4))
```
